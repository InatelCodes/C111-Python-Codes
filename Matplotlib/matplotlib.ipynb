Importações

```python
import matplotlib.pyplot as plt
import numpy as np
```

Plot

```python
x = np.array([1, 2, 3, 4])
y = x*2

plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Gráfico de linhas (plot)')

plt.plot(x, y, 'h:r', linewidth=3, markersize=10)
plt.show()
```

Plot com múltiplas linhas

```python
x = np.array([1, 2, 3, 4])
y = x*2
y2 = x*x

plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.title('Plot com múltiplas linhas')

plt.plot(x,y, 'g--o', x, y2,'r-.h')

plt.show()
```

Subplots

```python
plt.subplot(2, 2, 1)
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.plot(x, y, 'g--o')

plt.subplot(2, 2, 2)
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.plot(x, y2, 'r-s')

plt.subplot(2, 2, 3)
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.plot(x, y, 'b:*')

plt.subplot(2, 2, 4)
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.plot(x, y2, 'y-+')

plt.tight_layout()
plt.show()
```

Gráfico de Dispersão - Outlier

```python
import pandas as pd
```

```python
dfpaises = pd.read_csv('paises.csv', delimiter=';')

# os 6 maiores países em área
paises = dfpaises.nlargest(6, 'Area (sq. mi.)')

print(len(dfpaises))
print(len(paises))
```

```python
plt.scatter(paises['Country'], paises['Area (sq. mi.)'], s=paises['Area (sq. mi.)']/10000)

plt.show()
```

```python
plt.scatter(paises['Country'], paises['GDP ($ per capita)'], s=paises['Area (sq. mi.)']/10000)

plt.show()
```

Gráfico de Barras

```python
# os 5 maiores países em PIB per capita
paises = dfpaises.nlargest(5, 'GDP ($ per capita)')

nome = paises['Country']
valores = paises['GDP ($ per capita)']

plt.bar(nome, valores, color='pink')

plt.show()
```

Gráfico em Torta/Pizza

```python
filter= dfpaises['Coastline (coast/area ratio)']==0

paises = dfpaises[filter]

# Quantidade de paises que não possuem costa marítima
qtNoCoast = len(paises)
# Quantidade de países que possuem costa marítima
qtCoast = len(dfpaises) - qtNoCoast

# '%1.1f%%'
# primeira % -> indica que é um número
# o primeiro numero 1 -> a dimensão mínima "para não por espaço"
# o segundo e o terceiro são quantos items .2f -> define a quantidade de casas decimais
# os dois ultimos %% -> indica que é uma porcentagem

plt.pie([qtNoCoast, qtCoast], labels=['Sem costa marítima', 'Com costa marítima'], autopct='%1.1f%%', colors=['lightgray', 'lightblue'])

plt.show()
```
