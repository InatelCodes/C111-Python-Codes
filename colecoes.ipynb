Listas

```python
nomes = ['Goku', 'Vegeta', 'Gohan', 'Trunks']

print(nomes)
```

Inserindo dados na lista

```python
nomes.append('Bulma') # inserindo no final da lista
```

```python
nomes.insert(2, 'Piccolo') # inserindo na posição 2 da lista
```

```python
print(nomes)
```

```python
del nomes[3] # deletando o elemento na posição 3 da lista

print(nomes)
```

```python
nomes.pop(2) # deletando o elemento na posição 2 da lista

print(nomes)
```

```python
nomes.remove('Vegeta') # deletando o elemento 'Vegeta' da lista
print(nomes)
```

Adicionando tratamentos de erros

```python
nomes.remove('Vegeta') #deletando o mesmo elemento 'Vegeta' da lista novamente, mas ele não existe mais na lista, então vai dar erro
```

```python
# Evitando o erro acima
if 'Vegeta' in nomes:
    nomes.remove('Vegeta')
    print('Vegeta está na lista')
else:
    print('Vegeta não está na lista')
```

```python
tamanho_lista = len(nomes)
print(f'Tamanho da lista: {tamanho_lista}')

if tamanho_lista > 4:
    nomes.pop(4)
    print('Item removido pelo indice!')
else:
    print('Lista muito pequena, não removido')
```

Ordenação

```python
print(nomes)
```

```python
nomes.sort() # ordenando a lista em ordem alfabética
print(nomes)
```

```python
nomes.sort(reverse=True) # ordenando a lista em ordem alfabética reversa
print(nomes)
```

```python
nomes1 = sorted(nomes) # ordenando a lista em ordem alfabética sem alterar a lista original
print(nomes1)
```

```python
nomes1.append(True)
nomes1.append(47)
nomes1.append(3.14)
print(nomes1)
```

```python
print(nomes1[0])
nomes1[0] = 'Inatel'
print(nomes1[0])
print(nomes1)
```

Tuplas

```python
nomes_tupla = ('Goku', 'Vegeta', 'Gohan', 'Trunks')
print(nomes_tupla)
print(type(nomes_tupla))
print(type(nomes))
```

```python
print(nomes_tupla[0])
print(nomes_tupla[1:4]) # primeiro item é inclusivo e o último item é exclusivo, por isso 4 e não 3
```

```python
print(nomes_tupla)
# tentando alterar um item na tupla
nomes_tupla[2] = 'Bulma'
```

```python
for nome in nomes_tupla:
    print(nome)

print('-----------')

for count in range(len(nomes_tupla)):
    print(f'Índice: {count}, valor: {nomes_tupla[count]}')
```

Operações com Tuplas

```python
a = (2, 6, 8)
b = (4, 6, 9, 5)
z = a + b

print(z)

print(min(z), max(z)) # menor numero e o maior

print(z.count(6)) # numero de ocorrencias
```

Conjunto (Set)

```python
nomes_set = {'Goku', 'Vegeta', 'Gohan', 'Trunks', 'Gohan'} # ele não permite itens duplicados
print(nomes_set)
print(type(nomes_set))
```

```python
nomes_set.remove('Trunks')
nomes_set.add('TRUNKS')
print(nomes_set)
```

Operações com os conjuntos

```python
a = {2, 6, 8}
b = {1, 6, 5}

# União
z = a | b
print(f'União: {z}')

#Diferença
z = a - b
print(f'Diferença de A e B: {z}')
z = b - a
print(f'Diferença de B e A: {z}')

# Interseção
z = a & b
print(f'Interseção: {z}')
```

Dicionários

```python
dados = {
    "nome": ' Goku',
    "idade": 47,
    "temNome": True,

}

print(dados)

# Busca por chave especifica
print(dados['nome'])
```

```python
# Criar uma nova informação
dados['sexo'] = 'M'
print(dados)
```

```python
# Exclui uma informação
del dados['sexo']
print(dados)
```

```python
# Alterar informações
dados['idade'] = 30
print(dados)
```

```python
dados1 = {
    "nome": ' Vegeta',
    "idade": 40,
    "temNome": True,

}

dados2 = {
    "nome": ' Gohan',
    "idade": 35,
    "temNome": True,

}
```

```python
banco = []
banco.append(dados)
banco.append(dados1)
banco.append(dados2)

print(f"Banco: {banco}")
```

```python
for dicts in banco:
  print(dicts)
```
