```python
import pandas as pd
```

```python
# criando laabels, eqiuvalente aos índices de um vetor
labels = ['a', 'b', 'c'] # se não criar, atribui-se automaticamente os índices 0, 1, 2..

# criando uma lista de dados que queremos armazenar
data = [10, 20, 30]

serie1 = pd.Series(data, index=labels)

print(serie1)
```

```python
# acessando o valor especifico associado ao label 'a'
print(serie1['a'])
```

Operações com séries

```python
s1 = pd.Series({'a': 10, 'b': 20, 'c': 30})
s2 = pd.Series({'a': 10, 'c': 50, 'd': 80})
```

```python
# Soma os valores associados aos labels iguais, e atribui NaN aos labels que não existem em ambos
print(s1 + s2)
```

```python
# Soma os valores associados aos labels iguais, e atribui 0 aos labels que não existem em ambos
print(s1.add(s2, fill_value=0))
```

```python
print(serie1.min())
print(serie1.max())
print(serie1.mean())
```

Slices, fatiando a série

```python
print(serie1[['c', 'a']])
```

```python
# slicing, retorna os valores a partir do índice 1 (inclusive) até o final
print(serie1[1:])

# slicing, retorna os valores a partir do índice 0 (inclusive) até o índice 2 (exclusive)
print(serie1[0:2])
```

```python
# serie booleana, retorna True para os valores maiores que 25 e False para os valores menores ou iguais a 25
print(serie1 > 25)

print()

# apliquei a condição para ver dados maiores que 25
mascara = serie1 > 25
print(mascara)

print()

print(serie1[mascara])
```

```python
# condição mostra os números maiores que a média da série
print(serie1[serie1 > serie1.mean()])

# condição mostra os números que, quando divididos por 2, são iguais a 10
print(serie1[serie1/2 == 10])
```

Trabalhando com DataFrames

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(10)

df = pd.DataFrame(
    index = ['A', 'B', 'C', 'D', 'E'],
    columns = ['W', 'X', 'Y', 'Z'],
    data = np.random.randint(1, 50, [5,4])
)

# acessando dataframe
print(df)
```

```python
# acessando uma série "coluna"
print(df['W'])
# acessando um valor da série w
print(df['W']['A'])

print()

print(type(df))
# retorna o tipo da coluna W, que é uma série
print(type(df['W']))
```

```python
print(df[['Z', 'Y']])
```

```python
print(df.loc[['C', 'D'], ['X', 'Y']])
```

```python
print(df.iloc[2:4, 1:3])
```
