```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Scatterplot

```python
db_tips = sns.load_dataset('tips')
print(db_tips)
```

```python
sns.scatterplot(data=db_tips, x='total_bill', y='tip', hue='smoker')
plt.xlabel('Total da conta')
plt.ylabel('Gorjeta')

plt.show()
```

Histplot

```python
db_penguins = sns.load_dataset('penguins')
print(db_penguins)
```

```python
sns.histplot(data=db_penguins, x='flipper_length_mm', hue='species', kde=True)
```

Boxplot

```python
print(db_tips)
```

```python
sns.boxplot(data=db_tips, x='day', y='tip', hue='sex')

plt.show()
```

##### Regplot
Gráfico de dispersão com linha de regressão ajustada;
A parte sombreada indica a incerteza da estimativa da linha

```python
sns.regplot(data=db_tips, x='total_bill', y='tip', line_kws={'color': 'red'})
```

#### Heatmap
Dados em formato de matriz utilizando cores para indicar a intensidade ou magnitude dos valores

```python
corr = db_tips[['total_bill', 'tip', 'size']].corr()
print(corr)
```

```python
sns.heatmap(corr, annot=True, fmt='.2f')
```
