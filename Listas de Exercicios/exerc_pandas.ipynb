```python
import pandas as pd
import numpy as np
```

Questão 01:

```python
seriesAno1 = pd.Series({
    'Java': 16.25,
    'C': 16.04,
    'Python': 9.85
})

seriesAno2 = pd.Series({
    'C': 16.21,
    'Python': 12.12,
    'Java': 11.68
})

print('Series do Ano 1:')
print(seriesAno1)

print('\nSeries do Ano 2:')
print(seriesAno2)
```

Questão 02:

```python
totalAno1 = seriesAno1.sum()
totalAno2 = seriesAno2.sum()

print(f'Total do Ano 1: {totalAno1:.2f}%')
print(f'Total do Ano 2: {totalAno2:.2f}%')
```

Questão 03:

```python
crescimento = seriesAno2 - seriesAno1

print('Crescimento/declínio de cada linguagem:')
print(crescimento)
```

Questão 04:

```python
crescimentos = crescimento[crescimento > 0]

print('Linguagens que tiveram crescimento:')
print(crescimentos)
```

Questão 05:

```python
projecaoAno3 = seriesAno2 + crescimento
projecaoAno4 = projecaoAno3 + crescimento

print('Projeção para o Ano 4:')
print(projecaoAno4)

print('\nLinguagem mais popular:')
print(projecaoAno4.nlargest(1))
```

Questão 06:

```python
df = pd.DataFrame(
    index=['A', 'B', 'C', 'D', 'E'],
    columns=['W', 'X', 'Y', 'Z'],
    data=np.random.randint(1, 50, [5, 4])
)

print('DataFrame:')
print(df)

mediaX = df.loc[df['X'] < 30, 'X'].mean()

print(f'\nMédia dos elementos da coluna X menores que 30: {mediaX:.2f}')
```

Questão 07:

```python
mediaD = df.loc['D'].mean()

somaE = df.iloc[4].sum()

print(f'Média dos elementos da linha D: {mediaD:.2f}')
print(f'Soma dos elementos da linha E: {somaE}')
```

Questão 08:

```python
recorte = df.loc[['A', 'C', 'E'], ['X', 'Y']]

print('Slicing:')
print(recorte)

soma_linhas = recorte.sum(axis=1)

print('\nSoma dos elementos de cada linha:')
print(soma_linhas)

soma_colunas = recorte.sum(axis=0)

print('\nSoma dos elementos de cada coluna:')
print(soma_colunas)
```
