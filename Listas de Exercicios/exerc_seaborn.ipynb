```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

Questão 01

```python
iris = sns.load_dataset('iris')

setosa = iris[iris['species'] == 'setosa']
correlacao = setosa.select_dtypes(include='number').corr()

print(correlacao.round(2))

sns.heatmap(correlacao, annot=True, cmap='coolwarm',
            vmin=-1, vmax=1, fmt='.2f')
plt.title('Correlação entre variáveis - Setosa')
plt.show()
```

Questão 02

```python
titanic = sns.load_dataset('titanic')

sns.histplot(data=titanic, x='age', hue='sex',
             kde=True, bins=20, element='step')

plt.title('Distribuição das idades por sexo')
plt.xlabel('Idade')
plt.ylabel('Quantidade de passageiros')
plt.show()
```

Questão 03

```python
titanic = sns.load_dataset('titanic')

sns.boxplot(data=titanic, x='class', y='age', hue='sex')

plt.title('Idades por classe e sexo no Titanic')
plt.xlabel('Classe')
plt.ylabel('Idade')
plt.show()
```

Questão 04

```python
mpg = sns.load_dataset('mpg').dropna(
    subset=['horsepower', 'mpg']
)

sns.regplot(data=mpg, x='horsepower', y='mpg',
            scatter_kws={'alpha': 0.6})

plt.title('Potência x Consumo dos veículos')
plt.xlabel('Potência (horsepower)')
plt.ylabel('Consumo (mpg)')
plt.show()
```

Questão 05

```python
dfpaises = pd.read_csv('../seaborn/paises.csv', sep=';')

colunas = [
    'GDP ($ per capita)',
    'Literacy (%)',
    'Infant mortality (per 1000 births)',
    'Phones (per 1000)'
]

dados = dfpaises[colunas].apply(pd.to_numeric, errors='coerce')
correlacao = dados.corr()

print(correlacao.round(2))

sns.heatmap(correlacao, annot=True, cmap='coolwarm',
            vmin=-1, vmax=1, fmt='.2f')

plt.title('Correlação entre indicadores dos países')
plt.show()
```
