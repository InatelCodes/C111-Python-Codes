```python
import numpy as np
```

```python
spacex = np.loadtxt('../NumPy/src/space.csv', delimiter=';', dtype=str, encoding='utf-8', skiprows=1)
```

```python
print(spacex[0])
```

Questão 01:

```python
status_missao = spacex[:, 7]

sucessos = np.sum(status_missao == 'Success')
total_missoes = len(status_missao)

porcentagem = (sucessos / total_missoes) * 100

print(f'Porcentagem de missões que deram certo: {porcentagem:.2f}%')
```

Questão 02:

```python
custos = spacex[:, 6]

custos_disponiveis = custos[custos != '']
custos_disponiveis = custos_disponiveis.astype(float)

media = np.mean(custos_disponiveis)

print(f'Média dos gastos das missões: {media:.2f}')
```

Questão 03:

```python
localizacoes = spacex[:, 2]

missoes_eua = np.sum(
    np.char.find(localizacoes, 'USA') >= 0
)

print(f'Quantidade de missões realizadas pelos EUA: {missoes_eua}')
```

Questão 04:

```python
empresas = spacex[:, 1]
custos = spacex[:, 6]
detalhes = spacex[:, 4]

filtro_spacex = empresas == 'SpaceX'

custos_spacex = custos[filtro_spacex]
detalhes_spacex = detalhes[filtro_spacex]

filtro_custo = custos_spacex != ''

custos_spacex = custos_spacex[filtro_custo]
detalhes_spacex = detalhes_spacex[filtro_custo]

custos_spacex = custos_spacex.astype(float)

indice = np.argmax(custos_spacex)

print('Missão mais cara realizada pela SpaceX:')
print(f'Detalhes: {detalhes_spacex[indice]}')
print(f'Custo: {custos_spacex[indice]:.2f}')
```

Questão 05

```python
empresas = spacex[:, 1]

empresas_unicas, quantidades = np.unique(
    empresas,
    return_counts=True
)

print('Quantidade de missões por empresa:\n')

for empresa, quantidade in zip(empresas_unicas, quantidades):
    print(f'{empresa}: {quantidade} missões')
```

Questão  06

```python
status_foguete = spacex[:, 5]

aposentados = np.sum(status_foguete == 'StatusRetired')
total_missoes = len(status_foguete)

porcentagem = (aposentados / total_missoes) * 100

print(f'Porcentagem de missões com foguetes StatusRetired: {porcentagem:.2f}%')
```
