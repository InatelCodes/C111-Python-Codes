Questão 1:

```python
nome = input("Entre com seu nome completo: ")

# todas as letras maiúsculas
print(nome.upper())

# todas as letras minúsculas
print(nome.lower())

# qtas letras (sem espaços)
print(len(nome.replace(" ", "")))

# trocar o ultimo nome
troca_sobrenome = nome.rsplit(" ", 1)[0] + " do Inatel"
print(troca_sobrenome)
```

Questão 2:

```python
numero = int(input("Quer ver a tabuada de qual número? "))
inicio = int(input("Começar em qual número? "))
fim = int(input("Terminar em qual número? "))

print(f"\n--- Tabuada do {numero} (de {inicio} até {fim}) ---")

for i in range(inicio, fim + 1):
    resultado = numero * i
    print(f"{numero} x {i:2d} = {resultado}")
```

Questão 3:

```python
sexo = input("Qual o seu sexo? (M/F) ").strip().upper()

while sexo not in ["M", "F"]:
    print("Sexo inválido. Digite M para masculino ou F para feminino.")
    sexo = input("Qual o seu sexo? (M/F) ").strip().upper()

if sexo == "F":
    print("Você é do sexo feminino.")
else:
    print("Você é do sexo masculino.")
```

```python
distancia = float(input("Digite a distância em km:"))

preco = 0

if distancia <= 200:
    preco = distancia * 0.50
else:
    preco = distancia * 0.45

print(f"O preço da passagem é: R$ {preco:.2f}")
```

Questão 5:

```python
numero = int(input("Digite um número entre 1000 e 9999:"))

while numero < 1000 or numero > 9999:
    print("Número inválido. Digite um número entre 1000 e 9999.")
    numero = int(input("Digite um número entre 1000 e 9999:"))

unidade = numero % 10
dezena = (numero // 10) % 10
centena = (numero // 100) % 10
milhar = (numero // 1000) % 10

print(f'Número digitado: {numero}')
print(f'Unidade: {unidade}')
print(f'Dezena: {dezena}')
print(f'Centena: {centena}')
print(f'Milhar: {milhar}')
```

Questão 6:

```python
import math

num_decimal = float(input("Digite um número decimal: "))

raiz_quadrada = num_decimal ** 0.5 
funcao_teto = math.ceil(num_decimal)
funcao_chao = math.floor(num_decimal)
parte_inteira = int(num_decimal)

print(f"Número digitado: {num_decimal}")
print(f"Raiz quadrada: {raiz_quadrada}")
print(f"Função teto: {funcao_teto}")
print(f"Função chão: {funcao_chao}")
print(f"Parte inteira: {parte_inteira}")
```

Questão 7:

```python
palavra = input("Digite uma palavra: ")

# cada letra em maiusculo, uma por linha
for letra in palavra:
    print(letra.upper())

# quantas vogais
vogais = "AEIOUaeiou"
contador_vogais = 0
tem_a = False
for letra in palavra:
    if letra in vogais:
        contador_vogais += 1
    if letra == "a" or letra == "A":
        tem_a = True
print(f"Quantidade de vogais: {contador_vogais}")

if tem_a:
    print("A palavra contém a letra 'a'.")
else:
    print("A palavra não contém a letra 'a'.")
```

Questão 8:

```python
num_1 = int(input("Digite o primeiro número: "))
num_2 = int(input("Digite o segundo número: "))

# operações
adicao = num_1 + num_2
subtracao = num_1 - num_2
multiplicacao = num_1 * num_2
divisao = num_1/num_2
resto_divisao = num_1 % num_2
potencia = num_1 ** num_2

print(f"Números digitados: {num_1} e {num_2}")
print(f"Soma: {adicao} \nSubtração: {subtracao} \nMultiplicação: {multiplicacao} \nDivisão: {divisao} \nResto da divisão: {resto_divisao} \nPotência: {potencia}")

#adicao par ou impar
if adicao % 2 == 0:
    print(f"A soma {adicao} é par.")
else:
    print(f"A soma {adicao} é ímpar.")
```
