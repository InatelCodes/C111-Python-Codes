```python
# Criando a lista com os 5 primeiros colocados
times = ["São Paulo", "Bota Fogo", "Barcelona", "Real Madrid", "Manchester City"]

# Mostrando os 3 primeiros colocados
print("3 primeiros colocados:")
print(times[:3])

# Mostrando os últimos 2 colocados
print("\nÚltimos 2 colocados:")
print(times[-2:])

# Lista em ordem alfabética
print("\nTimes em ordem alfabética:")
print(sorted(times))

# Encontrando a posição do Barcelona
print("\nPosição do Barcelona:")
print(times.index("Barcelona") + 1) # para contar como 1, 2, 3.. e não 0, 1, 2 ..
```

Questão 4

```python
pessoas = []

for i in range(3):
    nome = input(f"Digite o nome da pessoa {i + 1}: ")
    peso = float(input(f"Digite o peso de {nome}: "))

    pessoa = {
        "nome": nome,
        "peso": peso
    }

    pessoas.append(pessoa)

mais_pesada = max(pessoas, key=lambda pessoa: pessoa["peso"])
mais_leve = min(pessoas, key=lambda pessoa: pessoa["peso"])

print("\nPessoa mais pesada:")
print(mais_pesada["nome"])

print("\nPessoa mais leve:")
print(mais_leve["nome"])
```

Questão 5

```python
n = int(input("Quantas pessoas serão cadastradas? "))

soma_idades = 0
mulheres_menores_20 = 0

for i in range(n):
    nome = input(f"\nDigite o nome da pessoa {i + 1}: ")
    idade = int(input("Digite a idade: "))
    sexo = input("Digite o sexo (M/F): ").upper()

    soma_idades += idade

    if sexo == "F" and idade < 20:
        mulheres_menores_20 += 1

media_idade = soma_idades / n

print("\nMédia de idade do grupo:", media_idade)
print("Quantidade de mulheres com menos de 20 anos:", mulheres_menores_20)
```

Questão 8

```python
produtos = []

for i in range(3):
    print(f"\nProduto {i + 1}")

    nome = input("Nome: ")
    preco = float(input("Preço: R$ "))
    quantidade = int(input("Quantidade em estoque: "))

    produto = {
        "nome": nome,
        "preco": preco,
        "quantidade": quantidade
    }

    produtos.append(produto)

print("\n--- ESTOQUE ---")

for produto in produtos:
    total = produto["preco"] * produto["quantidade"]

    print(f"Produto: {produto['nome']}")
    print(f"Valor total em estoque: R$ {total:.2f}")
    print()
```
