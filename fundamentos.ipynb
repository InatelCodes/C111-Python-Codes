Concatenações

```python
print(True, False)
print("PI=", 3.1415)
print("Pi=" + "3.1415")
print("PI={}, Euler={}".format(3.1415, 2.71828))
print("PI={0}, Euler={1}".format(3.1415, 2.71828))
print("PI={1}, Euler={0}".format(3.1415, 2.71828))
print(f'PI={3.1415}')

pi = f'\nPI={3.1415}'
num = "PI={1}, Euler={0}".format(3.1415, 2.71828)

print(pi)
print(num)
```

Entrada de Dados com o Teclado

```python
nome = input("Digite seu nome: ")
print(f'Olá {nome}, seja bem-vindo(a)!')

ano_nasc = int(input("Digite seu ano de nascimento: "))
idade = 2026 - ano_nasc
print(f'Você tem {idade} anos.')

peso = float(input("Digite seu peso (em kg): "))
print(f'Seu peso é {peso} kg.')

print("Nome, tipo: ", type(nome), "Conteudo: ", nome)
print("Idade, tipo: ", type(idade), "Conteudo: ", idade)
print("Peso, tipo: ", type(peso), "Conteudo: ", peso)
```

Operadores Aritméticos

```python
a = 10
b = 3

print('a=', a, 'b=', b)

# Operadores Aritméticos
print('\nSoma: ', a + b)
print('Subtração: ', a - b)
print('Multiplicação: ', a * b)
print('Divisão: ', a / b)
print('Potenciação: ', a ** b)
print('Divisão inteira: ', a // b)
print('Resto da divisão: ', a % b)

# Operadores de Comparação
print('\na == b: ', a == b)
print('a == 10: ', a == 10)

# Precedencia dos Operadores
print('\n2 + 3 * 4 = ', 2 + 3 * 4)
print('(2 + 3) * 4 = ', (2 + 3) * 4)
```

Importação de Bibliotecas

```python
import math

num = 2.3

print(math.ceil(num))
print(math.floor(num))
print(math.trunc(num))

print('Raiz quadrada de 16: ', math.sqrt(16))
print('Logaritmo natural de 10: ', math.log(10))
print('Coseno de 0: ', math.cos(0))
```

Manipulando Cadeias de Caracteres

```python
palavra = 'Hello World'

print('String', palavra)

print(palavra[6])
print(palavra[:5])
print(palavra[6:11])
print(palavra[3:])
print(palavra[::2]) # saltos
print(palavra[::-1]) # inverso

print('World' in palavra)

palavra = palavra.replace('World', 'Python')
print(palavra)
```

Estrutura de Decisão

```python
x = int(input("Digite um número inteiro: "))

if x < 0:
    print("Número negativo")
elif x == 0:
    print("Número zero")
else:
    print("Número positivo")
```

Estrutura de Repetição

```python
for i in range(10):
    print(i)

print('--------')

for i in range(-5, 10):
    print(i)

print('--------')

for i in range(10, 0, -1):
    print(i)

print('--------')

for i in range(0, 10, 2):
    print(i)

print('--------')

coisas = ['cachorro', True, 3, 2.71]
for i in coisas:
    print(i)
```
